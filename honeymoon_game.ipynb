```python
import random

print("🎉 ¡BIENVENIDOS AL JUEGO DE LUNA DE MIEL! 🎉\n")
print("=" * 70)
print("Este juego te ayudará a elegir el DESTINO PERFECTO para tu luna de miel")
print("basado en las preferencias de AMBOS miembros de la pareja.\n")
print("=" * 70)

# Phase 1: Proposal
q = "¿Quieres casarte conmigo?"
ans = input(q + ' ')
ans_norm = ans.strip().lower()

if ans_norm in ('sí','si','s','yes','y'):
    print('¡Sí! ¡Ahora a planificar la luna de miel! 💍\n')
elif ans_norm in ('no','n'):
    print("Has dicho 'no'. Cambiando respuesta a 'sí'...")
    print('¡Perfecto! ¡Ahora a planificar la luna de miel! 💍\n')
else:
    print('Respuesta no clara. Asumiré que es un "sí"')
    print('¡Ahora a planificar la luna de miel! 💍\n')

# Phase 2: Two-Player Preference Gathering
print("=" * 70)
print("FASE 1: PREFERENCIAS DE AMBOS\n")

# Initialize scores for each destination type
preference_scores = {
    'playa': 0,
    'ciudad': 0,
    'naturaleza': 0,
    'aventura': 0,
    'cultura': 0,
    'relax': 0
}

players = ["PERSONA 1", "PERSONA 2"]

for player in players:
    print(f"\n📋 Turno de {player}:")
    print("-" * 70)
    
    # Question 1: Vibe
    print("\n¿Cuál es tu vibe ideal?")
    print("  a) Playa, sol y piña colada 🏖️")
    print("  b) Cafés, museos y gastronomía 🍷")
    print("  c) Montañas, bosques y naturaleza 🌲")
    ans1 = input("Tu respuesta (a/b/c): ").strip().lower()
    if ans1 == 'a':
        preference_scores['playa'] += 2
        preference_scores['relax'] += 1
    elif ans1 == 'b':
        preference_scores['ciudad'] += 2
        preference_scores['cultura'] += 1
    elif ans1 == 'c':
        preference_scores['naturaleza'] += 2
        preference_scores['relax'] += 1
    
    # Question 2: Activity Level
    print("\n¿Prefieres?")
    print("  a) Relajarse y desconectar 😌")
    print("  b) Explorar y descubrir lugares nuevos 🗺️")
    print("  c) Hacer deportes y aventuras extremas 🪂")
    ans2 = input("Tu respuesta (a/b/c): ").strip().lower()
    if ans2 == 'a':
        preference_scores['relax'] += 2
        preference_scores['playa'] += 1
    elif ans2 == 'b':
        preference_scores['aventura'] += 2
        preference_scores['cultura'] += 1
    elif ans2 == 'c':
        preference_scores['aventura'] += 2
        preference_scores['naturaleza'] += 1
    
    # Question 3: Climate
    print("\n¿Cuál es tu clima favorito?")
    print("  a) Tropical y caluroso ☀️")
    print("  b) Templado y fresco 🌤️")
    print("  c) Frío extremo e invierno ❄️")
    ans3 = input("Tu respuesta (a/b/c): ").strip().lower()
    if ans3 == 'a':
        preference_scores['playa'] += 2
        preference_scores['relax'] += 1
    elif ans3 == 'b':
        preference_scores['ciudad'] += 1
        preference_scores['cultura'] += 1
    elif ans3 == 'c':
        preference_scores['aventura'] += 1
        preference_scores['naturaleza'] += 1
    
    # Question 4: Budget Mindset
    print("\n¿Cuál es tu actitud con el dinero?")
    print("  a) Ahorro máximo, queremos presupuesto 💰")
    print("  b) Equilibrio: disfrutar sin quebrar piggy bank 💳")
    print("  c) ¡YOLO! Lujo total, que se arruine después 💎")
    ans4 = input("Tu respuesta (a/b/c): ").strip().lower()
    if ans4 == 'a':
        preference_scores['naturaleza'] += 1
        preference_scores['aventura'] += 1
    elif ans4 == 'b':
        pass  # No bonus, equilibrado
    elif ans4 == 'c':
        preference_scores['ciudad'] += 1
        preference_scores['playa'] += 1

# Phase 3: Calculate Best Destination
print("\n" + "=" * 70)
print("ANALIZANDO PREFERENCIAS...")
print("=" * 70 + "\n")

# Destination database with preference matching
destinos_db = {
    'Maldivas': {
        'emoji': '🏝️',
        'desc': 'Islas privadas, arrecifes de coral, bungalós sobre el agua',
        'prefs': {'playa': 3, 'relax': 3, 'naturaleza': 1},
        'presupuesto': '$$$$'
    },
    'París': {
        'emoji': '🗼',
        'desc': 'Ciudad romántica, gastronomía, museos, Eiffel',
        'prefs': {'ciudad': 3, 'cultura': 3, 'playa': 0},
        'presupuesto': '$$$'
    },
    'Bali': {
        'emoji': '🧘',
        'desc': 'Templos, arrozales, playas, yoga y meditación',
        'prefs': {'naturaleza': 2, 'cultura': 2, 'relax': 2},
        'presupuesto': '$$'
    },
    'Islandia': {
        'emoji': '❄️',
        'desc': 'Glaciares, cascadas, auroras boreales, aventura extrema',
        'prefs': {'aventura': 3, 'naturaleza': 3, 'playa': 0},
        'presupuesto': '$$$'
    },
    'Costa Rica': {
        'emoji': '🌴',
        'desc': 'Selva tropical, playas, volcanes, zipline, vida silvestre',
        'prefs': {'aventura': 2, 'naturaleza': 3, 'playa': 1},
        'presupuesto': '$$'
    },
    'Italia (Amalfi)': {
        'emoji': '🍋',
        'desc': 'Costa mediterránea, pueblos pintorescos, pasta italiana',
        'prefs': {'ciudad': 2, 'playa': 2, 'cultura': 2},
        'presupuesto': '$$$'
    },
    'Tailandia': {
        'emoji': '🏯',
        'desc': 'Templos, playas tropicales, mercados, comida callejera',
        'prefs': {'ciudad': 1, 'cultura': 2, 'playa': 2, 'aventura': 1},
        'presupuesto': '$'
    },
    'Noruega': {
        'emoji': '⛰️',
        'desc': 'Fiordos, montañas, paisajes dramáticos, naturaleza virgen',
        'prefs': {'naturaleza': 3, 'aventura': 2, 'playa': 0},
        'presupuesto': '$$$$'
    }
}

# Calculate match scores for each destination
destination_matches = {}
for dest_name, dest_info in destinos_db.items():
    match_score = 0
    for pref_type, pref_value in dest_info['prefs'].items():
        match_score += pref_value * preference_scores[pref_type]
    destination_matches[dest_name] = match_score

# Get top 3 recommendations
top_3 = sorted(destination_matches.items(), key=lambda x: x[1], reverse=True)[:3]

print("🏆 TOP 3 DESTINOS RECOMENDADOS PARA USTEDES:\n")
for i, (dest_name, score) in enumerate(top_3, 1):
    dest = destinos_db[dest_name]
    print(f"{i}. {dest['emoji']} {dest_name.upper()}")
    print(f"   {dest['desc']}")
    print(f"   💵 Presupuesto: {dest['presupuesto']}")
    print(f"   ⭐ Compatibilidad: {'█' * int(score/5)} {score}/100\n")

# Final recommendation
print("=" * 70)
best_dest_name = top_3[0][0]
best_dest = destinos_db[best_dest_name]

print(f"✨ NUESTRO DESTINO GANADOR: {best_dest['emoji']} {best_dest_name.upper()} ✨\n")
print(f"¿Por qué? Este destino coincide perfectamente con las preferencias")
print(f"combinadas de ambos. ¡Empaquen maletas y disfruten! 🧳💑\n")

# Random honeymoon challenge
print("=" * 70)
print("PERO ESPERA... ¡TODO VIAJE TIENE SORPRESAS!\n")
desafios = [
    ("Tu pareja pierde el pasaporte → Vacaciones improvisadas en el hotel 🤦", "📋"),
    ("Se os olvida la cámara → Fotos solo mentales (las mejores) 📸", "📷"),
    ("El hotel no tiene la suite → Upgrade a cabaña con WiFi de cabra 📶", "🏨"),
    ("Lluvia tropical inesperada → ¡Más excusas para quedarse en la cama! 🌧️", "🌧️"),
    ("Os perdéis en la ciudad → Descubrís un restaurante secreto increíble 🍽️", "🗺️"),
    ("Se rompe la maleta → Compráis ropa local (¡Souvenir garantizado!) 👕", "🧳")
]

desafio_random = random.choice(desafios)
print(f"  {desafio_random[1]} {desafio_random[0]}\n")

print("=" * 70)
print("¡FELICIDADES POR VUESTRO MATRIMONIO Y LUNA DE MIEL! 🎊💍✈️")
print("=" * 70)
```
